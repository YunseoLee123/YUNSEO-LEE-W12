```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time
import pandas as pd
import os

driver = webdriver.Chrome()

# 🔥 기존 url 제거 → 메인 페이지로
driver.get("https://www.jobkorea.co.kr/")
time.sleep(3)

# 🔥 팝업 닫기
try:
    close_btn = driver.find_element(By.CSS_SELECTOR, "button[aria-label='닫기']")
    close_btn.click()
except:
    pass

time.sleep(1)

# 🔥 검색
search_box = driver.find_element(By.CSS_SELECTOR, "input")
search_box.send_keys("데이터분석")
search_box.send_keys(Keys.RETURN)

time.sleep(5)  # 로딩 대기

data = []

# 🔥 기존 그대로 유지
items = driver.find_elements(By.CSS_SELECTOR, "div.w-full.rounded-2xl")

for item in items:
    try:
        company = item.find_element(By.CSS_SELECTOR, "span a").text
    except:
        company = None

    try:
        title = item.find_element(By.CSS_SELECTOR, "span.truncate").text
    except:
        title = None

    try:
        link = item.find_element(By.CSS_SELECTOR, "a").get_attribute("href")
    except:
        link = None

    try:
        detail = item.find_element(By.CSS_SELECTOR, "div.flex.flex-col.gap-\\[10px\\]").text
    except:
        detail = None

    if not title:
        continue

    data.append({
        "Site": "Jobkorea",
        "Col_Company": company,
        "Col_Recruit": title,
        "Col_detail": detail,
        "Col_url": link
    })

df = pd.DataFrame(data)

os.makedirs("data_tmp", exist_ok=True)

df.to_csv("data_tmp/data_jobkorea.csv", index=False, encoding="utf-8-sig")

print(df.head())

driver.quit()
```
