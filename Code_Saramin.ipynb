```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time
import pandas as pd
import os

driver = webdriver.Chrome()

driver.get("https://www.saramin.co.kr/")
time.sleep(3)

# 🔥 검색창 먼저 열기
driver.find_element(By.ID, "btn_search").click()
time.sleep(1)

# 🔥 검색 입력
search_box = driver.find_element(By.ID, "ipt_keyword_recruit")
search_box.send_keys("데이터분석")
search_box.send_keys(Keys.RETURN)

time.sleep(3)

data = []

items = driver.find_elements(By.CSS_SELECTOR, "div.item_recruit")

for item in items:
    try:
        company = item.find_element(By.CSS_SELECTOR, "div.area_corp strong a").text
    except:
        company = None

    try:
        title = item.find_element(By.CSS_SELECTOR, "div.area_job a").text
    except:
        title = None

    try:
        link = item.find_element(By.CSS_SELECTOR, "div.area_job a").get_attribute("href")
    except:
        link = None

    # 🔥 핵심 수정 (제목 아래 조건 영역)
    try:
        detail = item.find_element(By.CSS_SELECTOR, "div.area_job div.job_condition").text
    except:
        detail = None

    data.append({
        "Site": "Saramin",
        "Col_Company": company,
        "Col_Recruit": title,
        "Col_detail": detail,
        "Col_url": link
    })

df = pd.DataFrame(data)

# 폴더 생성
os.makedirs("data_tmp", exist_ok=True)

# csv 저장
df.to_csv("data_tmp/data_saramin.csv", index=False, encoding="utf-8-sig")

print(df.head())

driver.quit()
```
